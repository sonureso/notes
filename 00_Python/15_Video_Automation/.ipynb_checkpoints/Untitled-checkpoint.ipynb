```python
from moviepy.editor import *
import moviepy.video.fx.all as fx
```

```python
def crop_img(img_obj, ratio):
    (w, h) = img_obj.size
    if(w/h > ratio):
        new_h = h
        new_w = h * ratio
    else:
        new_h = w * (1/ratio)
        new_w = w
    img_obj = fx.crop(img_obj, width=new_w, height=new_h, x_center=w/2, y_center=h/2)
    return img_obj
```

```python
# variables
clips = []
req_w = 1920
req_h = 1080

# reading images
clip1 = ImageClip('01_images/image1.jpg').set_duration(5)
clip2 = ImageClip('01_images/image2.jpg').set_duration(5)

# cropping in require ratio of width and hight
ratio = req_w/req_h
clip1 = crop_img(clip1, ratio)
clip2 = crop_img(clip2, ratio)

# sesizing in the required canvas size:
clip1 = clip1.resize((req_w,req_h))
clip2 = clip2.resize((req_w,req_h))
```

```python
video = CompositeVideoClip([clip1.crossfadeout(1), clip2.set_start(5).crossfadein(1.5)])
video.write_videofile("final.mp4", fps=24, remove_temp=True, codec="libx264", audio_codec="aac")
```

```python
def buy_now(data,obj):
    trx = {}
    obj.share_rate = data.close 
    obj.share_count = int(obj.curr_fund/data.close)
    obj.curr_fund = obj.curr_fund - obj.share_count*data.close
    trx['action'] = 'buy'
    trx['price'] = data.close
    trx['share_count'] = obj.share_count
    obj.transactions.append(trx)
    obj.next_action = 'sell'
    obj.value = obj.curr_fund + obj.share_count*data.close
    return obj

def sell_now(data,obj):
    trx = {}
    obj.share_rate = 0 
    obj.curr_fund = obj.curr_fund + obj.share_count*data.close
    obj.share_count = 0
    trx['action'] = 'sell'
    trx['price'] = data.close
    trx['share_count'] = obj.share_count
    obj.transactions.append(trx)
    obj.next_action = 'buy'
    obj.value = obj.curr_fund + obj.share_count*data.close
    return obj
```

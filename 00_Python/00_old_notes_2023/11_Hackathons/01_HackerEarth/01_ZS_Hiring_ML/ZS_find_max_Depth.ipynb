```python
def max_depth (N, edges, w):
    data = {}
    data['e'] = edges
    data['w'] = w
    data = del_node(data)
    count = len(data['w']) - data['w'].count(-1)
    depth = get_depth(data)
    return (count,depth-1)

def get_depth(data,d=0,layer_one=[1]):
    if(len(layer_one)==0):
        return d
    else:
        layer_two = [x[1] for x in data['e'] if x[0] in layer_one]
        return get_depth(data,d+1,layer_two)
    
def del_node(data):
    p = [x[0] for x in data['e']]
    r = [x[1] for x in data['e']]
    d = list(set(r)-set(p))
    to_del = []
    for x in d:
        if(data['w'][x-1]%2==0):
            to_del.append(x)
    if(len(to_del)!=0):
        new_edge = []
        for edge in data['e']:
            if(edge[1]==to_del[0]):
                pass
            else:
                new_edge.append(edge)
        data['e'] = new_edge
        data['w'][to_del[0]-1] = -1
        return del_node(data)
    else:
        return data
```

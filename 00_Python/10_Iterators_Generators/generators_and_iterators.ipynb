# 01.1 - Creating generator for 3 elements:

1. Generators: 
    a. generator returns an iterator on which you can iterate.
    b. You can iterate only one time over all elements.

```python
# Generator for yielding few values which we can further iterate over:
def gen(x):
    yield x**2
    yield x+2
    yield x+3
```

```python
# Creating an iterator:
obj = gen(25)

# iterating over it:
print("First Value: ",next(obj))
print("Second Value: ",obj.__next__())
print("Third Value: ",next(obj))
```

# 01.2 Creating Feconacy with Generators

```python
def feb(n):
    x = 0
    y = 1
    yield x
    yield y
    for i in range(n-2):
        yield x+y
        x,y=y,x+y
```

```python
f = feb(5)
for x in f:
    print(x)
```

# 01.3 - Generators Example-2

```python
def vowel():
    yield 'a'
    yield 'e'
    yield 'i'
    yield 'o'
    yield 'u'
```

```python
x = vowel()
for j in x:
    print(j)
```

# 02.2 - Creating my own iterator:

```python
class even():
    def __init__(self,max_val=20):
        self.val = 2
        self.max_val = max_val
    def __iter__(self):
        return self
    def __next__(self):
        if self.val<=self.max_val:    
            val = self.val
            self.val+=2
            return val
        else:
            raise StopIteration
```

```python
x = even(9)
for z in x:
    print(z)
```

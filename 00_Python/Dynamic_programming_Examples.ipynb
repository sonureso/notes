## 01-Longest increasing sequence problem

#####  Question Statement:
• Find the longest increasing subsequence:

```python
def lis(data):
    ans = [1] * len(data)
    for i in range(1,len(data)):
        ans[i] = 1 + max([ans[j] for j in range(i) if data[j]<data[i]], default=0)
    return max(ans)
```

```python
data = [3,1,8,2,5]
lis(data)
```

#### Question 2:
• Find the maximum height of stack of boxes where a box is represented by: (L1,W1,H1)

• A box2 can be stacked on another box1 iff:  L1<L2 and W1<W2. 

```python
def get_hos(boxes,box):
    hos = box[2] + max([get_hos(boxes,i) for i in boxes if(box[0]>i[0] and box[1]>i[1])], default=0)
    return hos
def get_max_hos(boxes):
    return max([get_hos(boxes,box) for box in boxes])
```

```python
boxes = [[4,5,3],[2,3,2],[3,6,2],[1,5,4],[2,4,1],[1,2,2]]
get_max_hos(boxes)
```

### 01. import libraries

```python
from moviepy.editor import *
import moviepy.video.fx.all as fx
import os
from time import process_time
from math import sin
```

### 02. Some helper functions

```python
def crop_img(img_obj, ratio):
    (w, h) = img_obj.size
    if(w/h > ratio):
        new_h = h
        new_w = h * ratio
    else:
        new_h = w * (1/ratio)
        new_w = w
    img_obj = fx.crop(img_obj, width=new_w, height=new_h, x_center=w/2, y_center=h/2)
    return img_obj

def is_image(name):
    if('.jpg' in name or '.png' in name or '.JPG' in name):
        return True
    else:
        return False
    
def create_clip(folder, logo, delay):
    clips = []
    skipped = []
    req_w = 1920.0
    req_h = 1080.0
    ratio = req_w/req_h
    fadeouttime = 1.5
    fadeintime = 1.5
    # reading images
    files = os.listdir(folder)
    files = [file for file in files if is_image(file)]

    for file in files:
        try:
            clips.append(ImageClip(folder+'/'+file))
        except:
            skipped.append(file)
    print("Total Images: ",len(clips))
    print("Skipped Images count: ",len(skipped))
    print(skipped)
    
    # set duration for each:
    clips = [clip.set_duration(delay+fadeintime) for clip in clips]

    # cropping in require ratio of width and hight
    clips = [crop_img(clip, ratio) for clip in clips]

    # sesizing in the required canvas size:
    clips = [clip.resize((req_w,req_h)) for clip in clips]

    # apply: crossfadein and crossfadeout:
    clips = [clip.crossfadein(fadeintime) for clip in clips]

    first_clip = clips[0].set_start(0).set_duration(delay)
    clips = [clip.set_start( clips.index(clip) * delay - fadeintime) for clip in clips[1:]]
    clips.insert(0, first_clip)
    
    if(logo['check'] == True):
        clips = add_logo(clips, logo['path'], delay)
    else:
        print("logo check is False! No logo added.")
    
    return clips

def add_logo(clips, logo_path = "", delay = 5):
    height = 85
    opacity = 0
    pos = ("right","top")
    margin_1 = 25
    margin_2 = 12

    logo_duration = delay * len(clips)
    logo = (ImageClip(logo_path).set_duration(logo_duration).resize(height=height) # if you need to resize...
              .margin(right=margin_1, top=margin_2, opacity=opacity) # (optional) logo-border padding
              .set_pos(pos))
    clips.append(logo)
    return clips

def save_clip(clips, path):
    video = CompositeVideoClip(clips)
    video.write_videofile(path, fps=24, remove_temp=True, codec="libx264", audio_codec="aac")
```

### 03. variables used

```python
path = "E:/06_Trendliner/02_Videos/28_feb/03_Content/"
folder = path+"images"
logo_path = path+"logo/logo_small.png"
logo = {'check':True, 'path': logo_path}
```

```python
clip = create_clip(folder,logo,delay=12)
#save_clip(clip, path+"final.mp4")
```

### 04. Saving Video

```python
%time
t1_start = process_time()
video = CompositeVideoClip(clip)
video.write_videofile(path+"version_01.mp4", fps=24, codec="libx264", audio_codec="aac")
t1_stop = process_time()
print("minutes taken:",(t1_stop-t1_start)/60.0) 
```

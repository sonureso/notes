```python
# adding some punctuation if char crosses 100 without any punctuation.
```

```python
def add_punc(st,a,b):
    #print(st,"\n")
    data = []
    for line in st.split("."):
        data.append(line.split(","))
    data2 = []
    for sentence in data:
        temp = []
        for line in sentence:
            if(len(line)>99):
                line = line.replace(a,b)
            temp.append(line)
        temp = ",".join(temp)
        data2.append(temp)
    return ".".join(data2)
```

```python
y = "He also suggested that the pandemic was intentionally created by those in power to control the population but whatevev is the case it is and knoth djsjd dsjds. This is, a new sentence."
y = add_punc(y," and ",", and ")
y = add_punc(y," but ",", but ")
y = add_punc(y," that ",", that ")
```

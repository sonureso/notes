```python
from hashlib import sha256
import time
```

```python
def hash_it(text):
    return sha256(text.encode('ascii')).hexdigest()
```

```python
nonce_limit = 10000000
zeroes = 6
def mine(block,trn,prev_hash):
    start = time.time()
    for nonce in range(nonce_limit):
        base_str = str(block)+str(trn)+prev_hash+str(nonce*2)
        hashed_str = sha256(base_str.encode('ascii')).hexdigest()
        if(hashed_str.startswith('0'*zeroes)):
            print("Total run time: ",time.time()-start)
            print("Great, we found the Hash at nonce: ",nonce*2)
            return hashed_str
    print("Not Found!!")
    return -1
```

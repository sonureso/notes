### 01. Helper Codes

```python
from moviepy.editor import *
from moviepy.video.fx.all import fx
```

### 00. Some functions

```python
def crop_img(img_obj, ratio):
    (w, h) = img_obj.size
    if(w/h > ratio):
        new_h = h
        new_w = h * ratio
    else:
        new_h = w * (1/ratio)
        new_w = w
    img_obj = fx.crop(img_obj, width=new_w, height=new_h, x_center=w/2, y_center=h/2)
    return img_obj
```

#### 01. Using: concatenate_videoclips

```python
clips = []
clip1 = ImageClip('01_images/image1.jpg').set_duration(5)
clip2 = ImageClip('01_images/image2.jpg').set_duration(5)
### transition: but this is slower.
clip1 = clip1.crossfadeout(1.0)
clip2 = clip2.crossfadein(2.0)
### transitions: faster
clip1 = moviepy.video.fx.all.fadeout(clip1, 1.0)
clip2 = moviepy.video.fx.all.fadein(clip2, 2.0, initial_color=None)
### Adding all clips
clips.append(clip1)
clips.append(clip2)
# Saving the video:
video = concatenate_videoclips(clips, method = 'compose')
video.write_videofile("final.mp4", fps=24, remove_temp=True, codec="libx264", audio_codec="aac")
```

##### 02. Using: CompositeVideoClips

```python
clips = []
clip1 = ImageClip('01_images/image1.jpg').set_duration(5)
clip2 = ImageClip('01_images/image2.jpg').set_duration(5)

#clip1 = clip1.set_position("center")
#clip2 = clip2.set_position("center")

clip1 = clip1.resize((1280,720))
clip2 = clip2.resize((1280,720))
```

```python
video = CompositeVideoClip([clip1.crossfadeout(1), # starts at t=0
                            clip2.set_start(5).crossfadein(1.5)])
video.write_videofile("final.mp4", fps=24, remove_temp=True, codec="libx264", audio_codec="aac")
```
